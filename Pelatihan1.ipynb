# Fungsi

```python
def salam():
  print('Halo, Selamat Pagi!')
def salam_nama(nama):
  print('Halo', nama+', Selamat Pagi!')
def salam_nama_kota(nama,kota):
  print(f'Halo {nama} di {kota}, Selamat Pagi!')
def salam_user(nama='Gamal',kota='Jakarta'):
  print(f'Halo {nama} di {kota}, Selamat Pagi!')
```

```python
salam_nama_kota('Gamal','Jakarta')
salam_nama('Gamal')
salam()
salam_user('Rio', 'Bogor')
```

```python
def tarif_pph_op(penghasilan):
  if penghasilan <= 60000000:
    return 0.05
  elif penghasilan > 60000000 and penghasilan <= 250000000:
    return 0.15
  elif penghasilan > 250000000:
    return 0.25
```
